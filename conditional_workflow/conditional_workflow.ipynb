```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict , Literal
```

```python
class BaseState(TypedDict):
    a : int
    b : int
    c: int

    equation : str
    discriminant : float
    result : str
```

```python
def show_equation(state: BaseState):
    equation = f'{state["a"]}x2{state["b"]}x{state["c"]}'

    return {'equation' : equation}
```

```python
def calculate(state:BaseState):
    result = state["b"]**2 - (4*state["a"]*state["c"])

    return {
        'discriminant' : result
    }
    
```

```python
def real_roots(state:BaseState):
    root1 = (-state["b"] + state['discriminant']**0.5)/(2*state["a"])
    root2 = (-state["b"] - state['discriminant']**0.5)/(2*state["a"])

    result = f"the rootes are {root1} and {root2}"

    return {

        "result" : result

    }
```

```python
def repeated_roots(state:BaseState):
    root1 = (-state["b"]) /(2*state["a"])
    

    result = f"only repeated root is {root1} "

    return {

        "result" : result

    }
```

```python
def no_real_roots(state:BaseState):
     result = f"no reall roots"

     return {
          "result" : result
     }
```

```python
def check_condition(state:BaseState)-> Literal["real_roots", "repeated_roots" , "no_real_roots"]:
    if state['discriminant'] > 0 :
        return "real_roots"
    elif state['discriminant'] == 0:
        return "repeated_roots"
    else:
        return"no_real_roots"
```

```python
graph = StateGraph(BaseState)

graph.add_node('show_equation' , show_equation)
graph.add_node('cal_discriminant' , calculate )
graph.add_node('real_roots', real_roots)
graph.add_node('repeated_roots', repeated_roots)
graph.add_node('no_real_roots', no_real_roots)

graph.add_edge(START, 'show_equation')
graph.add_edge('show_equation' , 'cal_discriminant')
graph.add_conditional_edges('cal_discriminant', check_condition)
graph.add_edge('real_roots', END)
graph.add_edge('repeated_roots',END)
graph.add_edge('no_real_roots', END)


app = graph.compile()
```

```python
initial_state={
    'a' : 5,
    'b': 7,
    'c' : 9
}

app.invoke(initial_state)
```
