```python
from langgraph.graph import StateGraph,START,END
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
llm = HuggingFaceEndpoint(
    repo_id="meta-llama/Llama-3.1-8B-Instruct",
    task="text-generation"
) 

model = ChatHuggingFace(llm = llm)

 
```

```python

class Blogstate(TypedDict):
    title : str
    outline : str
    content : str
```

```python
def create(state : Blogstate) -> Blogstate:
    title = state['title']

    prompt = f"Generate an outline for a blog on the topic {title}"

    result = model.invoke(prompt).content

    state['outline'] = result

    return state
```

```python
def create_blog(state: Blogstate)-> Blogstate:
    title = state['title']
    outline = state['outline']

    prompt = f"write a detailed blog on the title {title} using the following outline \n {outline}"

    blog_out = model.invoke(prompt).content

    state['content'] = blog_out

    return state
```

```python
graph = StateGraph(Blogstate)

graph.add_node('vivaan' , create)
graph.add_node('utke', create_blog)

graph.add_edge(START, 'vivaan')
graph.add_edge('vivaan' , 'utke')
graph.add_edge('utke', END)

workflow = graph.compile()
```

```python
initial_state = {'title' : 'AI will impact on human'}
final_state = workflow.invoke(initial_state)
print(final_state['content'])
```
